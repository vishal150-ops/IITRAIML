```python
import pandas as pd
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"

# Create DataFrames

married = pd.DataFrame({
    "emp_id": (1,2,3,4,5,6,7,8),
    "Married": ['John','James','Harper','Amy','Mia','Liz','Benjamin','Ava']
})
homeowners = pd.DataFrame({
    "emp_id": (9,10,11,12,5,6,7,8),
    "Homeowners": ['Sophia','William','Jacob','Michael','Mia','Liz','Benjamin','Ava']
})

#Sorting
married_sorted = married.sort_index(axis=1,ascending=False)
homeowners_sorted = homeowners.sort_values(by='emp_id', ascending=True)

print(married_sorted)
print()
print(homeowners_sorted)



```

```python
consolidated=pd.merge(married,homeowners,on='emp_id', how='inner') #try different inner,outer, left, right joins
consolidated.sort_index(axis=1,ascending=False) #axis=0 means row labels, #acis =1 means column labels
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
iris = load_iris()
iris
```

```python
df = pd.DataFrame (iris.data, columns = iris.feature_names)
df.head()
```

```python
df['species'] = iris.target
df.head()
```

```python
df['species']= df['species'].map ({
    0: 'Sentosa',
    1:'Versicolor',
    2:'Virginica'
})
df.head()
```

```python
#1 Bar chat
#***********
species_count = df['species'].value_counts()
species_count.plot (kind='bar')
plt.title("Bar Chart")
plt.xlabel("Species")
plt.ylabel("Count")
plt.show()
```

```python
# PieChart
species_count.plot(kind='pie',autopct='%1.1f%%')
plt.title("Pie Chart")
plt.ylabel("")
plt.show()
```

```python
#3. line graph
plt.plot(df['sepal length (cm)'])
plt.title("Line Chart")
plt.show()
```

```python
# 4 BOXPLOT
# ====================================
sns.boxplot(x='species',
            y='sepal length (cm)',
            data=df)
plt.title("Boxplot")
plt.show()
```

```python
corr = df.corr(numeric_only=True)
sns.heatmap(corr,annot=True)
plt.title("Heatmap")
plt.show()
```

```python
#6 PAir plot
sns.pairplot(df, hue='species')
plt.show()
```

```python
# 7 STACKED BAR CHART
# ====================================
table = pd.crosstab(df['species'],
                    df['petal length (cm)'] > 3)
table.plot(kind='bar',
           stacked=True)
plt.title("Stacked Bar Chart")
plt.show()
```

```python
# 8 CLUSTERED BAR CHART

# ====================================

sns.countplot(x='species',

              hue='species',

              data=df)

plt.title("Clustered Bar Chart")

plt.show()
```

```python
# ==========================================

# 2 HISTOGRAM

# Distribution of petal length

# ==========================================

plt.hist(df['petal length (cm)'])

plt.title("Histogram")

plt.xlabel("Petal Length")

plt.ylabel("Frequency")

plt.show()
```
